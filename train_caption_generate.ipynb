```python
from google.colab import files
from PIL import Image
import io

uploaded = files.upload()
image_path = next(iter(uploaded))
image = Image.open(io.BytesIO(uploaded[image_path])).convert("RGB")
image.show()
```

```python
from transformers import BlipProcessor, BlipForConditionalGeneration
import torch

# Load BLIP model and processor
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to("cuda" if torch.cuda.is_available() else "cpu")
```

```python
def generate_caption(image):
    inputs = processor(image, return_tensors="pt").to("cuda" if torch.cuda.is_available() else "cpu")
    out = model.generate(**inputs)
    caption = processor.decode(out[0], skip_special_tokens=True)
    return caption

caption = generate_caption(image)
print("🖼️ Caption:", caption)
```

```python
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
from gtts import gTTS
import torch
import tempfile
import os

device = "cuda" if torch.cuda.is_available() else "cpu"
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

def generate_caption_with_audio(image):
    image = image.convert("RGB")
    inputs = processor(image, return_tensors="pt").to(device)
    out = model.generate(**inputs)
    caption = processor.decode(out[0], skip_special_tokens=True)

    # Text-to-Speech using gTTS
    tts = gTTS(text=caption)
    temp_audio = tempfile.NamedTemporaryFile(delete=False, suffix=".mp3")
    tts.save(temp_audio.name)

    return caption, temp_audio.name
```

```python
import gradio as gr

iface = gr.Interface(
    fn=generate_caption_with_audio,
    inputs=gr.Image(type="pil", label="Upload an Image"),
    outputs=[
        gr.Textbox(label="Generated Caption"),
        gr.Audio(label="Voice Caption")
    ],
    title="🖼️ Image Caption Generator + Voice 🎙️",
    description="Upload an image and get a caption — both text and audio! Powered by BLIP + gTTS."
)

iface.launch(share=True)
```
